```python
#Function for cleaning column names

def clean_column_names(dataframe):
    """
    Return a copy of a dataframe with cleaned column names.

    Cleaning includes:
    - removing leading and trailing spaces;
    - replacing line breaks with spaces;
    - replacing repeated spaces with a single space.
    """

    cleaned_dataframe = dataframe.copy()

    cleaned_dataframe.columns = (
        cleaned_dataframe.columns
        .str.replace("\n", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

    return cleaned_dataframe


#Apply the function to both datasets

la_df = clean_column_names(la_df)
lea_df = clean_column_names(lea_df)

print("Column names cleaned successfully.")
```

```python
def dataset_summary(dataframe, dataset_name):
    """
    Produce a general summary of a dataframe.
    """

    summary = pd.DataFrame({
        "Column": dataframe.columns,
        "Data Type": dataframe.dtypes.astype(str).values,
        "Non-Missing": dataframe.notna().sum().values,
        "Missing": dataframe.isna().sum().values,
        "Unique Values": dataframe.nunique(dropna=False).values
    })

    print(f"{dataset_name} summary")
    print("-" * 60)
    print(f"Rows: {dataframe.shape[0]}")
    print(f"Columns: {dataframe.shape[1]}")

    return summary
```

```python
def missing_value_report(dataframe):
    """
    Return a missing-value report for a dataframe.
    """

    missing_count = dataframe.isna().sum()
    missing_percentage = (
        missing_count / len(dataframe)
    ) * 100

    report = pd.DataFrame({
        "Missing Values": missing_count,
        "Missing Percentage": missing_percentage
    })

    return report.sort_values(
        by="Missing Values",
        ascending=False
    )
```

```python
def convert_to_numeric(dataframe, columns):
    """
    Convert selected columns to numeric values.

    Values that cannot be converted become missing values,
    allowing them to be identified during validation.
    """

    converted_dataframe = dataframe.copy()

    for column in columns:
        if column in converted_dataframe.columns:
            converted_dataframe[column] = pd.to_numeric(
                converted_dataframe[column],
                errors="coerce"
            )

    return converted_dataframe
```

```python
def negative_value_report(dataframe, columns):
    """
    Count negative values in selected numerical columns.
    """

    report = {}

    for column in columns:
        report[column] = (dataframe[column] < 0).sum()

    return (
        pd.Series(report, name="Negative Values")
        .to_frame()
        .sort_values("Negative Values", ascending=False)
    )
```

```python
def rate_range_report(dataframe, rate_columns):
    """
    Report the minimum, maximum and number of invalid
    observations for rate variables expected to lie
    between zero and one.
    """

    rows = []

    for column in rate_columns:
        invalid_count = (
            (dataframe[column] < 0) |
            (dataframe[column] > 1)
        ).sum()

        rows.append({
            "Variable": column,
            "Minimum": dataframe[column].min(),
            "Maximum": dataframe[column].max(),
            "Outside 0–1 Range": invalid_count
        })

    return pd.DataFrame(rows)
```

```python
def safe_divide(numerator, denominator):
    """
    Divide two pandas Series safely.

    A zero denominator is replaced with a missing value
    to prevent division-by-zero errors.
    """

    denominator = denominator.replace(0, np.nan)

    return numerator / denominator
```

```python
#Percentage of the population aged 0–15

la_analysis["Child Population Percentage"] = (
    safe_divide(
        la_analysis[
            "Total Dependent child Population (0-15)"
        ],
        la_analysis["Total Population"]
    ) * 100
)


#Number of schools for every 10,000 pupils

la_analysis["Schools per 10,000 Pupils"] = (
    safe_divide(
        la_analysis["Total no of Schools"],
        la_analysis["Total no of pupils"]
    ) * 10000
)


#Number of special schools for every 10,000 pupils

la_analysis["Special Schools per 10,000 Pupils"] = (
    safe_divide(
        la_analysis["No of special schools"],
        la_analysis["Total no of pupils"]
    ) * 10000
)


#Pupils as a percentage of the total population

la_analysis["Pupils as Percentage of Population"] = (
    safe_divide(
        la_analysis["Total no of pupils"],
        la_analysis["Total Population"]
    ) * 100
)


print("LA derived variables created successfully.")
```

```python
#Percentage of the population classified as dependent

lea_analysis["Dependent Population Percentage"] = (
    safe_divide(
        lea_analysis["LEA Total Dependant Population"],
        lea_analysis["LEA Total Population"]
    ) * 100
)


#Number of schools for every 10,000 students

lea_analysis["Schools per 10,000 Students"] = (
    safe_divide(
        lea_analysis["LEA Total No of Schools"],
        lea_analysis["LEA Total Student Population"]
    ) * 10000
)


#Number of special schools for every 10,000 students

lea_analysis["Special Schools per 10,000 Students"] = (
    safe_divide(
        lea_analysis["LEA Total Special Schools"],
        lea_analysis["LEA Total Student Population"]
    ) * 10000
)


#Total DSG funding per student

lea_analysis["DSG per Student"] = (
    safe_divide(
        lea_analysis["Total DSG"],
        lea_analysis["LEA Total Student Population"]
    )
)


#High Needs Block funding per student

lea_analysis["High Needs Funding per Student"] = (
    safe_divide(
        lea_analysis["High Needs Block"],
        lea_analysis["LEA Total Student Population"]
    )
)


print("LEA derived variables created successfully.")
```

```python
def coefficient_of_variation(dataframe, columns):
    """
    Calculate the coefficient of variation for selected variables.

    CV = standard deviation / mean

    A larger value indicates greater relative variation.
    """

    results = []

    for column in columns:
        mean_value = dataframe[column].mean()
        standard_deviation = dataframe[column].std()

        cv = standard_deviation / mean_value

        results.append({
            "Variable": column,
            "Mean": mean_value,
            "Standard Deviation": standard_deviation,
            "Coefficient of Variation": cv
        })

    return pd.DataFrame(results)
```

```python
#LA relative variation

la_cv = coefficient_of_variation(
    la_analysis,
    LA_OUTCOMES
)

print("Relative variation in LA SEND outcomes:")
display(la_cv.round(4))

#LEA relative variation

lea_cv = coefficient_of_variation(
    lea_analysis,
    LEA_OUTCOMES
)

print("Relative variation in LEA SEND outcomes:")
display(lea_cv.round(4))
```

2.6.1 Create an IQR outlier function

```python
def identify_iqr_outliers(
    dataframe,
    outcome_column,
    authority_column
):
    """
    Identify observations outside 1.5 times the interquartile range.
    """

    q1 = dataframe[outcome_column].quantile(0.25)
    q3 = dataframe[outcome_column].quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - (1.5 * iqr)
    upper_limit = q3 + (1.5 * iqr)

    outliers = dataframe[
        (dataframe[outcome_column] < lower_limit) |
        (dataframe[outcome_column] > upper_limit)
    ][
        [
            authority_column,
            outcome_column
        ]
    ].copy()

    outliers["Lower IQR Limit"] = lower_limit
    outliers["Upper IQR Limit"] = upper_limit

    outliers["Outlier Direction"] = np.where(
        outliers[outcome_column] < lower_limit,
        "Low outlier",
        "High outlier"
    )

    return outliers.sort_values(
        outcome_column,
        ascending=False
    )
```

```python
#Identify LA outliers for each SEND outcome

la_outlier_tables = {}

for outcome in LA_OUTCOMES:

    outlier_table = identify_iqr_outliers(
        dataframe=la_analysis,
        outcome_column=outcome,
        authority_column="LA Authority"
    )

    la_outlier_tables[outcome] = outlier_table

    print(f"\nPotential outliers for {outcome}:")
    print(f"Number of potential outliers: {len(outlier_table)}")

    display(outlier_table.round(4))

#Identify LEA outliers for each SEND outcome

lea_outlier_tables = {}

for outcome in LEA_OUTCOMES:

    outlier_table = identify_iqr_outliers(
        dataframe=lea_analysis,
        outcome_column=outcome,
        authority_column="LEA (Local Education Authority)"
    )

    lea_outlier_tables[outcome] = outlier_table

    print(f"\nPotential outliers for {outcome}:")
    print(f"Number of potential outliers: {len(outlier_table)}")

    display(outlier_table.round(4))

```

```python
def rank_authorities(
    dataframe,
    authority_column,
    outcome_column,
    number=10
):
    """
    Return the highest and lowest authorities for an outcome.
    """

    ranking_data = dataframe[
        [
            authority_column,
            outcome_column
        ]
    ].copy()

    ranking_data["Percentage"] = (
        ranking_data[outcome_column] * 100
    )

    highest = (
        ranking_data
        .nlargest(number, outcome_column)
        .reset_index(drop=True)
    )

    lowest = (
        ranking_data
        .nsmallest(number, outcome_column)
        .reset_index(drop=True)
    )

    return highest, lowest
```

Calculate highest-to-lowest gaps

```python
def outcome_range_summary(
    dataframe,
    outcome_columns
):
    """
    Summarise the observed range of each outcome.
    """

    results = []

    for outcome in outcome_columns:

        minimum = dataframe[outcome].min()
        maximum = dataframe[outcome].max()

        results.append({
            "Outcome": outcome,
            "Minimum Percentage": minimum * 100,
            "Maximum Percentage": maximum * 100,
            "Percentage-Point Gap": (
                maximum - minimum
            ) * 100,
            "Maximum-to-Minimum Ratio": (
                maximum / minimum
                if minimum != 0
                else np.nan
            )
        })

    return pd.DataFrame(results)
```

Compare Pearson and Spearman correlations with the outcomes

```python
def outcome_correlation_table(
    pearson_matrix,
    spearman_matrix,
    outcomes
):
    """
    Create a focused table comparing Pearson and Spearman
    correlations between predictors and outcomes.
    """

    results = []

    for outcome in outcomes:

        for predictor in pearson_matrix.columns:

            if predictor not in outcomes:

                results.append({
                    "Outcome": outcome,
                    "Predictor": predictor,
                    "Pearson Correlation":
                        pearson_matrix.loc[
                            predictor,
                            outcome
                        ],
                    "Spearman Correlation":
                        spearman_matrix.loc[
                            predictor,
                            outcome
                        ]
                })

    result_table = pd.DataFrame(results)

    result_table["Absolute Pearson"] = (
        result_table[
            "Pearson Correlation"
        ].abs()
    )

    return result_table.sort_values(
        [
            "Outcome",
            "Absolute Pearson"
        ],
        ascending=[
            True,
            False
        ]
    )
```

```python
la_outcome_correlations = outcome_correlation_table(
    pearson_matrix=la_pearson_correlation,
    spearman_matrix=la_spearman_correlation,
    outcomes=LA_OUTCOMES
)

display(
    la_outcome_correlations[
        [
            "Outcome",
            "Predictor",
            "Pearson Correlation",
            "Spearman Correlation"
        ]
    ].round(3)
)
```

Identify strongly correlated predictor pairs

```python
def strong_correlation_pairs(
    correlation_matrix,
    threshold=0.70
):
    """
    Return unique variable pairs with absolute correlation
    equal to or greater than the selected threshold.
    """

    pairs = []

    columns = correlation_matrix.columns

    for first_index in range(len(columns)):

        for second_index in range(
            first_index + 1,
            len(columns)
        ):

            first_variable = columns[first_index]
            second_variable = columns[second_index]

            correlation = correlation_matrix.loc[
                first_variable,
                second_variable
            ]

            if abs(correlation) >= threshold:

                pairs.append({
                    "Variable 1": first_variable,
                    "Variable 2": second_variable,
                    "Correlation": correlation,
                    "Absolute Correlation": abs(correlation)
                })

    return (
        pd.DataFrame(pairs)
        .sort_values(
            "Absolute Correlation",
            ascending=False
        )
        if pairs
        else pd.DataFrame(
            columns=[
                "Variable 1",
                "Variable 2",
                "Correlation",
                "Absolute Correlation"
            ]
        )
    )
```

##Similar-Deprivation Authority Comparison

This part of the analysis is particularly relevant to the postcode-lottery argument.

It identifies authorities with similar deprivation levels but substantially different SEND outcomes.

Also to calculate differences within deprivation quintiles

```python
def within_group_outcome_ranges(
    dataframe,
    group_column,
    outcome_columns
):
    """
    Calculate the minimum, maximum and range of each outcome
    within each group.
    """

    results = []

    for group_name, group_data in dataframe.groupby(
        group_column,
        observed=False
    ):

        for outcome in outcome_columns:

            minimum = group_data[outcome].min()
            maximum = group_data[outcome].max()

            results.append({
                "Group": group_name,
                "Outcome": outcome,
                "Number of Authorities": len(group_data),
                "Minimum Percentage": minimum * 100,
                "Maximum Percentage": maximum * 100,
                "Within-Group Gap": (
                    maximum - minimum
                ) * 100
            })

    return pd.DataFrame(results)
```

```python
la_within_imd_ranges = within_group_outcome_ranges(
    dataframe=la_analysis,
    group_column="IMD Quintile",
    outcome_columns=LA_OUTCOMES
)

print(
    "Variation in SEND outcomes within deprivation quintiles:"
)

display(
    la_within_imd_ranges.round(2)
)
```

Identify the highest and lowest authority within each quintile

```python
def group_extreme_authorities(
    dataframe,
    group_column,
    authority_column,
    outcome_column
):
    """
    Identify the highest and lowest authority for an outcome
    within each analytical group.
    """

    results = []

    for group_name, group_data in dataframe.groupby(
        group_column,
        observed=False
    ):

        highest_index = group_data[
            outcome_column
        ].idxmax()

        lowest_index = group_data[
            outcome_column
        ].idxmin()

        highest_row = dataframe.loc[
            highest_index
        ]

        lowest_row = dataframe.loc[
            lowest_index
        ]

        results.append({
            "Group": group_name,
            "Highest Authority":
                highest_row[authority_column],
            "Highest Percentage":
                highest_row[outcome_column] * 100,
            "Lowest Authority":
                lowest_row[authority_column],
            "Lowest Percentage":
                lowest_row[outcome_column] * 100,
            "Percentage-Point Gap":
                (
                    highest_row[outcome_column] -
                    lowest_row[outcome_column]
                ) * 100
        })

    return pd.DataFrame(results)
```

```python
for outcome in LA_OUTCOMES:

    print("=" * 70)
    print(
        f"Highest and lowest authorities within "
        f"each IMD quintile: {outcome}"
    )

    quintile_extremes = group_extreme_authorities(
        dataframe=la_analysis,
        group_column="IMD Quintile",
        authority_column="LA Authority",
        outcome_column=outcome
    )

    display(
        quintile_extremes.round(2)
    )
```

Concise Automatic EDA Summary

```python
def print_eda_summary(
    dataframe,
    outcomes,
    dataset_name
):
    """
    Print a concise summary of outcome variation.
    """

    print("=" * 70)
    print(f"{dataset_name} EXPLORATORY SUMMARY")
    print("=" * 70)

    for outcome in outcomes:

        mean_percentage = (
            dataframe[outcome].mean() * 100
        )

        minimum_percentage = (
            dataframe[outcome].min() * 100
        )

        maximum_percentage = (
            dataframe[outcome].max() * 100
        )

        gap = (
            maximum_percentage -
            minimum_percentage
        )

        skewness = stats.skew(
            dataframe[outcome].dropna()
        )

        print(f"\n{outcome}")
        print(f"Mean: {mean_percentage:.2f}%")
        print(f"Minimum: {minimum_percentage:.2f}%")
        print(f"Maximum: {maximum_percentage:.2f}%")
        print(f"Observed gap: {gap:.2f} percentage points")
        print(f"Skewness: {skewness:.3f}")
```
